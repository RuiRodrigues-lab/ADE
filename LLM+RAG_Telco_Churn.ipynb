```python
from groq import Groq

client = Groq(api_key="HAHAHHAA, achas que eu caiu nesta? tá no vault, boa sorte! :P")

SYSTEM_PROMPT = """
És um especialista em Customer Relationship Management (CRM).
Respondes de forma técnica, objetiva e focada em estratégias de gestão de clientes, retenção e churn.
Responde em pt-pt.
"""

def ask_llm(prompt: str, rag_document: str = "", model: str = "llama-3.3-70b-versatile") -> str:
    messages = [
        {"role": "system", "content": SYSTEM_PROMPT},
    ]

    # Simulação de RAG (documento injetado no contexto)
    if rag_document:
        messages.append({
            "role": "system",
            "content": f"Contexto (RAG):\n{rag_document}"
        })

    messages.append({
        "role": "user",
        "content": prompt
    })

    response = client.chat.completions.create(
        model=model,
        messages=messages,
        temperature=0.2,
        max_completion_tokens=1024,
        top_p=1,
        stream=False
    )

    return response.choices[0].message.content
```

```python
doc = """
Cliente classificado como HIGH CHURN RISK.

Motivos:
- Tenure muito baixo (2 meses)
- Contrato mensal (sem fidelização)
- Baixa adoção de serviços adicionais (sem suporte técnico, proteção, streaming)
- Receita acumulada baixa (108.15)
- Serviço base apenas (DSL)

Perfil histórico semelhante a clientes que cancelam cedo por falta de engagement.

Ações recomendadas:
- Oferta de retenção (desconto temporário ou upgrade gratuito)
- Contacto proativo nos próximos 7 dias
- Sugerir bundle de serviços (TechSupport + DeviceProtection)
"""

prompt = """
Com base no perfil do cliente e no risco de churn, recomenda a melhor ação de retenção.
Simplifica a resposta para o operador apenas com ações recomendadas.
Mete a resposta em varias linhas por cada ação.
"""

resposta = ask_llm( prompt, rag_document=doc)

print("Ação Recomendada pela IA: " + resposta)
```
